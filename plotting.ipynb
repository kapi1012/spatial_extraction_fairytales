```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path
```

```python
def plot_loss(history):
    plt.figure(figsize=(8, 5))

    plt.plot(
        history["epoch"],
        history["train_loss"],
        marker="o",
        label="Training Loss"
    )

    plt.xlabel("Epoch")
    plt.ylabel("Loss")
    plt.title("Training Loss over Epochs")

    plt.grid(True)
    plt.legend()

    plt.show()
```

```python
def plot_val_metrics(history):
    plt.figure(figsize=(8, 5))

    plt.plot(
        history["epoch"],
        history["val_precision"],
        marker="o",
        label="Precision"
    )

    plt.plot(
        history["epoch"],
        history["val_recall"],
        marker="o",
        label="Recall"
    )

    plt.plot(
        history["epoch"],
        history["val_f1"],
        marker="o",
        label="F1"
    )

    best_epoch = history.loc[
        history["val_f1"].idxmax(),
        "epoch"
    ]

    best_f1 = history["val_f1"].max()

    plt.scatter(
        best_epoch,
        best_f1,
        s=100,
        label=f"Best F1={best_f1:.3f}"
    )

    plt.xlabel("Epoch")
    plt.ylabel("Score")
    plt.title("Validation Metrics")

    plt.ylim(0, 1)
    plt.grid(True)
    plt.legend()

    plt.show()
```

```python
report_path = Path("data/train-reports/training_history_2026_08_03-16_50_03.csv")

history = pd.read_csv(report_path)

history.head()
plot_loss(history)
```

```python
plot_val_metrics(history)
```
